# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
# Escalar la variable 'sepal length (cm)'

minmax_scaler = MinMaxScaler()
standard_scaler = StandardScaler()

data_minmax = minmax_scaler.fit_transform(data)
data_standard = standard_scaler.fit_transform(data)

# Crear gráfica de distribución antes y después del escalado

plt.figure(figsize=(12, 5))

# Distribución original
plt.subplot(1, 3, 1)
plt.hist(data['sepal length (cm)'], bins=10, edgecolor='black')
plt.title('Antes del escalado')
plt.xlabel('Sepal length (cm)')
plt.ylabel('Frecuencia')

# Distribución con MinMaxScaler
plt.subplot(1, 3, 2)
plt.hist(data_minmax[:, 0], bins=10, edgecolor='black')
plt.title('Después de MinMaxScaler')
plt.xlabel('Valor escalado')
plt.ylabel('Frecuencia')

# Distribución con StandardScaler
plt.subplot(1, 3, 3)
plt.hist(data_standard[:, 0], bins=10, edgecolor='black')
plt.title('Después de StandardScaler')
plt.xlabel('Valor estandarizado')
plt.ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = standard_scaler.fit_transform(data)

# Gráfica de dispersión de Sepal Length vs Sepal Width
plt.figure(figsize=(8, 6))

plt.scatter(data_standard[:, 0], data_standard[:, 1])

plt.title('Sepal Length vs Sepal Width - StandardScaler')
plt.xlabel('Sepal Length (estandarizado)')
plt.ylabel('Sepal Width (estandarizado)')

plt.show()
```
